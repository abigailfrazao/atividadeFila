# Implementação de Fila

# Nó da fila

```python
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None
```

# Fila

```python
class Queue:
    def __init__(self):
        self.front = None
        self.rear = None
    def enqueue(self, value):
        new_node = Node(value)

        if self.front == None:
            self.front = self.rear = new_node
            return
        self.rear.next = new_node
        self.rear = new_node
    # Remoção de elemnto da fila
    def dequeue(self):
        if self.front == None:
            raise Exception("Queue is empty")
        temp = self.front
        self.front = self.front.next
        if self.front == None:
            self.rear = None
        return temp.value

    # Acessa o primeiro elemento da fila
    def peek(self):
        if self.front == None:
            raise Exception("Queue is empty")
        return self.front.value
    
    # Verifica se a fila está vazia
    def is_empty(self):
        return self.front == None

    # Exibe a fila
    def display_queue(self):
        current = self.front
        while current != None:
            print(current.value, end=" ")
            current = current.next
```

# Teste de uso da fila

```python
fila = Queue()

fila.enqueue(10)
fila.enqueue(20)
fila.enqueue(30)

fila.display_queue()

# Desenfileirando dois elementos da fila
elemento_removido_1 = fila.dequeue()
print(f"Elemento removido: {elemento_removido_1}")
elemento_removido_2 = fila.dequeue()
print(f"Elemento removido: {elemento_removido_2}")

# Exibindo o estado da fila após os desenfileiramentos
print("Fila após desenfileirar dois elementos:")
fila.display_queue()

# Visualizando o primeiro elemento da fila
primeiro_elemento = fila.peek()
print(f"O primeiro elemento da fila é: {primeiro_elemento}")

# Verificando se a fila está vazia
if fila.is_empty():
    print("Fila está vazia")
else:
    print("Fila não está vazia")
```
